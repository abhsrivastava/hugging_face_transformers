```python
# Perform Imports

import gradio as gr
from transformers import pipeline

print(f'✅ Imports complete!')
```

```python
# Now we will load the fine tuned model we just published to HF

classifier = pipeline('text-classification', model="abhishes/novapay-sentiment")
sentence = "This product is amazing!!"
test_result = classifier(sentence)[0]
print(f'✅ Classifier is working')
print(f'Test Result for {sentence} -> {test_result['label']} ({test_result["score"]:.4f})')
```

```python
# Now let us write a simple Gradio App

def classify_text(text):
    result = classifier(text)[0]
    return f'{result["label"]} ({result["score"]:.4f})'


demo = gr.Interface(
    fn=classify_text,
    inputs="text",
    outputs="text"
)

demo.launch()
```
