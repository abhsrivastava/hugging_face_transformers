```python
# Imports

from transformers import pipeline
from transformers import AutoTokenizer

print(f"✅ Finished Imports")
```

```python
# Let's try the Chat Template instead of Raw Text

tokenizer = AutoTokenizer.from_pretrained("HuggingFaceTB/SmolLM2-1.7B-Instruct")

if tokenizer.chat_template is not None:
  print("This model expects a chat template")
else:
  print("Raw Text is fine!")

print("Pad token:", tokenizer.pad_token)
print("Pad token ID:", tokenizer.pad_token_id)


#Right Way
messages = [
  [
      {"role": "system", "content": "you are a customer support classifier!"},
      {"role": "user", "content": "Help me classify this customer message. please!"}
  ],
  [
      {"role": "system", "content": "you are a customer support classifier!"},
      {"role": "user", "content": "Tokenization is a crucial preprocessing step."}
  ],
  [
      {"role": "system", "content": "you are a customer support classifier!"},
      {"role": "user", "content": "Short text."}
  ]
]

tokenizer.padding_side = "left"
batch = tokenizer.apply_chat_template(messages,
                                      tokenize=True,
                                      add_generation_prompt=True,
                                      padding="longest",
                                      truncation=True,
                                      return_tensors="pt",
                                      return_dict=True)

print(batch["input_ids"].shape)
print(batch["attention_mask"].shape)

input_ids = batch["input_ids"]
attention_mask = batch["attention_mask"]

print(input_ids)
print(attention_mask)
print(attention_mask.sum(dim=1))
```
