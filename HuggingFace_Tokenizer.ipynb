```python
# Import packages
from transformers import AutoTokenizer

print("✅ Imports completed Successfully!")
```

```python
# Let us load the Auto Tokenizer and print some tokenizer attributes

CHECKPOINT = "distilbert-base-uncased"
tokenizer = AutoTokenizer.from_pretrained(CHECKPOINT)
print(f"Checkpoint: {CHECKPOINT}")
print(f"Tokenizer Class: {type(tokenizer).__name__}")
print(f"Vocab Size: {tokenizer.vocab_size}: tokens")
print(f"Model Max Length: {tokenizer.model_max_length}: tokens")
print(f"Pad Token: {tokenizer.pad_token} ")
print(f"Pad Token Id: {tokenizer.pad_token_id}")
```

```python
# Now let us tokenize a sentence

sentence = "NovaPay's Customer Support Team needs help!"

encoded = tokenizer(sentence)

print(f"Original sentence: {sentence!r}")
print(f"Number of tokens: {len(encoded["input_ids"])}")
print()
print(f"Input Ids: {encoded["input_ids"]}")
print(f"Attention Mask {encoded["attention_mask"]}")

```

```python
# Now let us try to decode the tokens

token_ids = encoded["input_ids"]
tokens = tokenizer.convert_ids_to_tokens(token_ids)
special_ids = tokenizer.all_special_ids
print(f"{'Position':<10} {'Token ID':<10} {'Token String:':<20} {'Notes'}")
print("-" * 65)
for i, (tid, tstr) in enumerate(zip(token_ids, tokens)):
  notes = ""
  if tid in special_ids:
    notes = "special token"
  elif tstr.startswith("##"):
    notes = "subword_continuation"
  print(f"{i:<10} {tid:<10} {tstr:<20} {notes}")


decoded_clean = tokenizer.decode(token_ids, skip_special_tokens=True)
print(f"Decoded clean: {decoded_clean!r}")
print(f"Original: {sentence!r}")
print(f"Round Trip Preserves the text content: {decoded_clean.strip() == sentence.lower()}")

```
